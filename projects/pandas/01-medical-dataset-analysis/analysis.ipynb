```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Sample medical dataset
sample_data = {
    'image_id': [
        'IMG001', 'IMG002', 'IMG003',
        'IMG004', 'IMG005', 'IMG006',
        'IMG007', 'IMG008', 'IMG008'
    ],
    'Pneumonia': [1, 0, 1, 0, 0, 1, 0, 0, 0],
    'Edema': [0, 1, 1, 0, 0, 0, 1, 0, 0],
    'Cardiomegaly': [0, 0, 0, 1, 0, 0, 0, 1, 1]
}

df = pd.DataFrame(sample_data)

df.loc[8, 'Pneumonia'] = None

df.to_csv('medical_lables.csv', index=False)

df.head()
```

```python
# Inspect the dataset shape
print("Dataset Shape: ", df.shape)
print("Number of Rows: ", df.shape[0])
print("Number of Columns: ", df.shape[1])
```

```python
# Check missing values
print(df.isnull().sum())

print()

# Calculate the percentage of missing values
missing_percentage = (df.isnull().sum() / len(df)) * 100
print("Percentage of missing values: ", missing_percentage)
```

```python
# Check duplicate records
print("Duplicate Rows: ", df.duplicated().sum())
```

```python
# Duplicate Image IDs
print("Duplicat Image IDs: ", df["image_id"].duplicated().sum())
```

```python
# Count positive disease labels
disease_columns = [
    'Pneumonia',
    'Edema',
    'Cardiomegaly'
]

positive_counts = df[disease_columns].sum()

print("Positive Disease Counts:")
print(positive_counts)
```

```python
# Calculate class distributions
for col in disease_columns:
    print("\nDisease: ", col)
    
    print(df[col].value_counts(dropna=False))
```

```python
# Visualize disease distribution
positive_counts.plot(
    kind='bar',
    figsize=(8, 5),
    title='Positive Medical Label Distribution'
)

plt.xlabel("Disease")
plt.ylabel("Number of Positive Samples")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Key Findings

1. The dataset contains 9 records and 4 columns.
2. One Pneumonia label is missing.
3. Duplicate image identifiers were detected.
4. Disease-positive counts were calculated.
5. Disease-label distributions were visualized.

## Conclusion

This exploratory analysis demonstrates the importance
of checking dataset structure, missing values, duplicate
records and class imbalance before machine learning.
